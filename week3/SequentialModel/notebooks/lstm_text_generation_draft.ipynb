# LSTM Text Generation Draft

Notebook nay dung de phan tich nhanh corpus va ghi lai y tuong LSTM text generation truoc khi refactor thanh pipeline trong `src/`.

````python
from pathlib import Path
from collections import Counter
import pandas as pd

PROJECT_ROOT = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
CORPUS_DIR = PROJECT_ROOT / 'data' / 'output'
files = sorted(CORPUS_DIR.glob('*.txt'))
len(files), files[:3]
````

````python
sample_files = files[:20]
rows = [{'file': path.name, 'chars': len(path.read_text(encoding='utf-8', errors='ignore'))} for path in sample_files]
pd.DataFrame(rows).describe()
````

````python
text = '\n\n'.join(path.read_text(encoding='utf-8', errors='ignore')[:5000] for path in sample_files)
counter = Counter(text)
print('sample chars:', len(text))
print('vocab size:', len(counter))
print('top chars:', counter.most_common(20))
````

## Model idea

Baseline trong `src/` la char-level LSTM:

```text
characters -> ids -> embedding -> LSTM -> linear -> next character
```

Char-level phu hop de draft nhanh vi khong can tokenizer, xu ly duoc dau tieng Viet va vocab nho. Sau khi baseline on co the nang cap sang word-level LSTM hoac Transformer.
